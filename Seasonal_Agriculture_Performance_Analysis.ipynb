```python
# ============================================
# STEP 21: CHECK FOR POTENTIAL OUTLIERS
# ============================================

# Calculate the number of potential outliers
# in each numerical column using the IQR method

outlier_summary = []

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Potential Outliers": len(outliers)
    })

# Convert results into a DataFrame
outlier_summary = pd.DataFrame(outlier_summary)

# Display the results
display(outlier_summary)
```

```python
# ============================================
# STEP 22: DISTRIBUTION OF NUMERICAL VARIABLES
# ============================================

# Create histograms for all numerical columns

df[numerical_columns].hist(
    figsize=(16, 12),
    bins=30
)

plt.suptitle(
    "Distribution of Numerical Variables",
    fontsize=16,
    y=1.02
)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# STEP 23: CORRELATION ANALYSIS
# ============================================

# Calculate the correlation matrix
correlation_matrix = df[numerical_columns].corr()

# Display the correlation matrix
display(correlation_matrix)
```

```python
# ============================================
# STEP 24: CORRELATION HEATMAP
# ============================================

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# ============================================
# STEP 25: IDENTIFY STRONGEST CORRELATIONS
# ============================================

# Create a copy of the correlation matrix
corr_pairs = correlation_matrix.copy()

# Remove the diagonal values (correlation of a variable with itself)
np.fill_diagonal(corr_pairs.values, np.nan)

# Convert the matrix into pairs of variables
strong_correlations = (
    corr_pairs
    .stack()
    .reset_index()
)

strong_correlations.columns = ["Variable 1", "Variable 2", "Correlation"]

# Remove duplicate pairs
strong_correlations["Pair"] = strong_correlations.apply(
    lambda row: tuple(sorted([row["Variable 1"], row["Variable 2"]])),
    axis=1
)

strong_correlations = strong_correlations.drop_duplicates(
    subset="Pair"
).drop(columns="Pair")

# Calculate absolute correlation for ranking
strong_correlations["Absolute Correlation"] = (
    strong_correlations["Correlation"].abs()
)

# Sort from strongest to weakest
strong_correlations = strong_correlations.sort_values(
    "Absolute Correlation",
    ascending=False
)

# Display the top 10 strongest correlations
print("Top 10 Strongest Correlations:")
display(strong_correlations.head(10))
```

```python
# Step 109: Statistical Significance of Top Yield Factors

top_5_factors = (
    correlation_with_yield
    .abs()
    .sort_values(ascending=False)
    .head(5)
    .index
)

significance_results = []

for factor in top_5_factors:
    data = df[[factor, "Yield_Tonnes_Ha"]].dropna()

    correlation, p_value = pearsonr(
        data[factor],
        data["Yield_Tonnes_Ha"]
    )

    significance_results.append({
        "Factor": factor,
        "Correlation": correlation,
        "P-value": p_value,
        "Statistically Significant": "Yes" if p_value < 0.05 else "No"
    })

significance_results = pd.DataFrame(significance_results)

print("Statistical Significance of Top 5 Yield Factors")
print("-" * 65)

display(significance_results)
```
